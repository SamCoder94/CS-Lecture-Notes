<p>For any comparison based sorting algorithm A, and any natural number n, there is an input of length n which requires A to perform Omega(nlogn) comparisons</p>
<p>Is it possible that there is an algorithm which performs better than nlogn?</p>
<p>It must not be comparison based, some examples are bucket sort and radix sort</p>
<h3>Bucket Sort</h3>
<p>Sort n numbers taking values in [0, k-1]</p>
<p>We have k buckets labelled 0 to k-1, putting elements into bucket labelled with their value, then emptying the buckets in order</p>
<p>We put them in a bucket by incrementing the value held in the corresponding bucket, which then says how many elements of that value there are</p>
<p>This works for values, but if we are sorting keys thent we treat the buckets as queues</p>
<p>This algorithm has O(n + k) worst case. Average case represented by O(n)
<p>It may be O(k) if the range is significantly larger than n. Further more it may be true that k is omega(nlogn) or omega(n^2)</p>
<hr>
<h3>Radix Sort</h3>
<p>We look 'One level below' the item values</p>
<p>We have a number of passes depending on the number of digits in the maximum value, and a number of buckets depending on the counting system (base 10 versus base 2)</p>
<p>Typically sort right most digit first</p>
<p>This sort works if the bucket sort stages are stable, meaning work done in previous rounds is not destroyed</p>
<p>If there are d bucket sort stages we have overall time complexity of O(dn), linear</p>
<p>d = log10(k) where k is taken from bucket sort, meaning we have O(nlog10(k))
<p>Radix Sort > Bucket sort with regards to space where k is non trivial</p>
<p> If k is linear function of n, Bucket sort is better, otherwise radix sort is better</p>
